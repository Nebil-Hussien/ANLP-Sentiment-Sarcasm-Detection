```python
import pandas as pd
from sklearn.metrics import f1_score, confusion_matrix
import matplotlib.pyplot as plt
```

```python
original = pd.read_csv("task_A_En_test.csv")
reannotated = pd.read_csv("task_A_test_shuffle_to_annotate.csv")
```

```python
reannotated
```

```python
merged = original.merge(reannotated, on = "text", suffixes = ("_original", "_reannotated"))

merged = merged[merged["sarcastic_reannotated"] != "?"]

merged["sarcastic_reannotated"] = merged["sarcastic_reannotated"].astype("int64")

merged
```

```python
confusion_matrix(merged["sarcastic_original"].array, merged["sarcastic_reannotated"].array)
```

```python
f1_score(merged["sarcastic_original"].array, merged["sarcastic_reannotated"].array, average = None)
```
